```python
# ============================================================
# CELL 2 — Full phishing email detection pipeline
# ============================================================

# ---------- Imports ----------
from google.colab import userdata
from huggingface_hub import login

login(token=userdata.get('HF_TOKEN'))

from datasets import load_dataset
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.metrics import classification_report, confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt
import re
import joblib
import anthropic

# ---------- Step 1: Load & split dataset ----------
dataset = load_dataset('zefang-liu/phishing-email-dataset')
df_train = pd.DataFrame(dataset['train'])
df_train, df_test = train_test_split(df_train, test_size=0.2, random_state=42)
df_train = df_train.reset_index(drop=True)
df_test  = df_test.reset_index(drop=True)

df_train = df_train.rename(columns={'Email Text': 'text', 'Email Type': 'label'})
df_test  = df_test.rename(columns={'Email Text': 'text', 'Email Type': 'label'})

df_train['label'] = (df_train['label'] == 'Phishing Email').astype(int)
df_test['label']  = (df_test['label']  == 'Phishing Email').astype(int)

# ---------- Step 2: Clean & build pipeline ----------
df_train = df_train.dropna(subset=['text']).reset_index(drop=True)
df_test  = df_test.dropna(subset=['text']).reset_index(drop=True)
df_train['text'] = df_train['text'].astype(str)
df_test['text']  = df_test['text'].astype(str)

model = Pipeline([
    ('tfidf', TfidfVectorizer(
        max_features=15000,
        ngram_range=(1, 2),
        stop_words='english',
        sublinear_tf=True
    )),
    ('clf', LogisticRegression(
        max_iter=1000, C=1.0, solver='lbfgs', n_jobs=-1
    ))
])

model.fit(df_train['text'], df_train['label'])
print("Model training complete.")

# ---------- Step 3: Evaluate ----------
y_pred = model.predict(df_test['text'])
y_true = df_test['label']

print(classification_report(y_true, y_pred, target_names=['Legitimate', 'Phishing']))

cm = confusion_matrix(y_true, y_pred)
ConfusionMatrixDisplay(cm).plot()
plt.show()

# ---------- Step 4: Hybrid scoring function ----------
def hybrid_phishing_score(email_text, ml_model):
    ml_prob  = float(ml_model.predict_proba([email_text])[0][1])
    ml_score = ml_prob * 60

    rule_score = 0
    text_lower = email_text.lower()
    rules = {
        'urgent action required': 10,  'verify your account': 10,
        'click here': 8,               'password': 7,
        'bank account': 8,             'limited time': 6,
        'congratulations you won': 10, 'suspend': 7,
        'unusual activity': 8,
    }
    for phrase, score in rules.items():
        if phrase in text_lower:
            rule_score += score
    if re.search(r'http[s]?://', text_lower):
        rule_score += 10
    rule_score = min(rule_score, 40)

    total_score = ml_score + rule_score
    if total_score >= 60:   verdict = 'PHISHING'
    elif total_score >= 35: verdict = 'SUSPICIOUS'
    else:                   verdict = 'LEGITIMATE'

    return {
        'ml_probability': round(ml_prob, 3),
        'ml_score': round(ml_score, 1),
        'rule_score': round(rule_score, 1),
        'total_score': round(total_score, 1),
        'verdict': verdict
    }

# ---------- Step 5: Save model ----------
joblib.dump(model, "phishing_model.joblib")
print("Model saved to phishing_model.joblib")

# ---------- Step 6: Gemini AI explainer ----------
client = anthropic.Anthropic(api_key=userdata.get('ANTHROPIC_API_KEY'))

def explain_verdict(email_text, score_result):
    prompt = f"""
    An email was analysed by a phishing detection system and scored as: {score_result['verdict']}
    ML probability: {score_result['ml_probability']}
    Rule-based score: {score_result['rule_score']}

    Email content:
    {email_text[:500]}

    In 2-3 sentences, explain to a non-technical user why this email
    was flagged or cleared. Be specific about what triggered the score.
    """
    message = client.messages.create(
        model="claude-sonnet-4-6",
        max_tokens=200,
        messages=[{"role": "user", "content": prompt}]
    )
    return message.content[0].text

# ---------- Step 7: Test with a sample email ----------
sample_email = """
Dear customer, your account has been suspended.
Urgent action required — click here to verify your password immediately.
http://suspicious-link.com
"""

import time

score = hybrid_phishing_score(sample_email, model)
print("\n--- Scores ---")
print(score)

time.sleep(2)  # wait 100 seconds before calling Gemini

explanation = explain_verdict(sample_email, score)
print("\n--- AI Explanation ---")
print(explanation)

from google.colab import files
files.download('phishing_model.joblib')
```

```python
# ============================================================
# CELL 3 — Process emails from teammate's Excel file
# ============================================================
import pandas as pd
import time
import os
from google.colab import files

# Upload emails.xlsx from your PC
print("Upload your emails.xlsx file:")
uploaded = files.upload()

def process_email_inbox(file_path='emails.xlsx'):
    df = pd.read_excel(file_path)
    print(f"Loaded {len(df)} emails.")

    # Add result columns
    df['verdict']     = None
    df['ml_score']    = None
    df['rule_score']  = None
    df['total_score'] = None
    df['explanation'] = None

    for idx, row in df.iterrows():
        subject    = str(row.get('Subject', ''))
        body       = str(row.get('Body', ''))
        email_text = subject + ' ' + body

        # ML scoring
        score = hybrid_phishing_score(email_text, model)

        # AI explanation
        explanation = explain_verdict(email_text, score)

        # Write results
        df.at[idx, 'verdict']     = score['verdict']
        df.at[idx, 'ml_score']    = score['ml_score']
        df.at[idx, 'rule_score']  = score['rule_score']
        df.at[idx, 'total_score'] = score['total_score']
        df.at[idx, 'explanation'] = explanation

        print(f"\nEmail {idx + 1} — From: {row.get('Sender', 'Unknown')}")
        print(f"Subject : {subject[:60]}")
        print(f"Verdict : {score['verdict']} (score: {score['total_score']})")
        print(f"Explanation: {explanation}")

        time.sleep(2)  # delay between Claude API calls

    # Save results to new Excel file
    output_path = 'phishing_results.xlsx'
    df.to_excel(output_path, index=False)
    print(f"\nDone. Results saved to {output_path}")

    # Download the results file automatically
    files.download(output_path)
    return df

results = process_email_inbox('emails.xlsx')
```
